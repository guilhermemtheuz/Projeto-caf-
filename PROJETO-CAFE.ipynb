```python
class Produto:

    def __init__(self, codigo, nome, categoria, preco, estoque):
        self.codigo = codigo
        self.nome = nome
        self.categoria = categoria
        self.preco = preco
        self.estoque = estoque


    def vender(self, quantidade):
        """
        Remove produtos do estoque quando uma venda é realizada
        """
        if quantidade <= self.estoque:
            self.estoque -= quantidade
            return True
        else:
            return False


    def adicionar_estoque(self, quantidade):
        """
        Adiciona produtos ao estoque
        """
        self.estoque += quantidade

    def calcular_valor_venda(self, quantidade):
        """
        Calcula o valor total de uma venda para o produto
        """
        return quantidade * self.preco

    def exibir(self):
        """
        Retorna os dados do produto
        """
        return (
            f"Código: {self.codigo} | "
            f"Produto: {self.nome} | "
            f"Categoria: {self.categoria} | "
            f"Preço: R$ {self.preco:.2f} | "
            f"Estoque: {self.estoque}"
        )

class Cliente:

    def __init__(self, codigo, nome, telefone):
        self.codigo = codigo
        self.nome = nome
        self.telefone = telefone


    def exibir(self):
        """
        Retorna os dados do cliente
        """
        return (
            f"Código: {self.codigo} | "
            f"Nome: {self.nome} | "
            f"Telefone: {self.telefone}"
        )

class Pedido:

    contador = 1


    def __init__(self, cliente):

        self.codigo = Pedido.contador
        Pedido.contador += 1

        self.cliente = cliente
        self.produtos = []
        self.total = 0



    def adicionar_produto(self, produto, quantidade):

        if produto.vender(quantidade):

            item = {
                "produto": produto,
                "quantidade": quantidade
            }

            self.produtos.append(item)

            self.total += produto.preco * quantidade

            return True

        return False



    def exibir(self):

        texto = "\n------ PEDIDO ------\n"

        texto += f"Pedido Nº: {self.codigo}\n"
        texto += f"Cliente: {self.cliente.nome}\n\n"


        for item in self.produtos:

            texto += (
                f"{item['produto'].nome} "
                f"x{item['quantidade']} "
                f"- R$ {item['produto'].preco * item['quantidade']:.2f}\n"
            )


        texto += f"\nTotal: R$ {self.total:.2f}\n"

        texto += "--------------------"

        return texto


class SistemaCoffeeShop:


    def __init__(self):

        self.produtos = []
        self.clientes = []
        self.pedidos = []



    # ===============================
    # PRODUTOS
    # ===============================


    def cadastrar_produto(self, produto):

        self.produtos.append(produto)



    def listar_produtos(self):

        if len(self.produtos) == 0:

            print("Nenhum produto cadastrado.")

        else:

            print("\n===== PRODUTOS =====")

            for produto in self.produtos:

                print(produto.exibir())



    def buscar_produto(self, codigo):

        for produto in self.produtos:

            if produto.codigo == codigo:

                return produto


        return None



    # ===============================
    # CLIENTES
    # ===============================


    def cadastrar_cliente(self, cliente):

        self.clientes.append(cliente)



    def listar_clientes(self):

        if len(self.clientes) == 0:

            print("Nenhum cliente cadastrado.")

        else:

            print("\n===== CLIENTES =====")

            for cliente in self.clientes:

                print(cliente.exibir())



    def buscar_cliente(self, codigo):

        for cliente in self.clientes:

            if cliente.codigo == codigo:

                return cliente


        return None



    # ===============================
    # PEDIDOS
    # ===============================


    def criar_pedido(self, cliente):

        pedido = Pedido(cliente)

        self.pedidos.append(pedido)

        return pedido



    def listar_pedidos(self):

        if len(self.pedidos) == 0:

            print("Nenhum pedido realizado.")

        else:

            for pedido in self.pedidos:

                print(pedido.exibir())



    # ===============================
    # RELATÓRIO
    # ===============================


    def faturamento(self):

        total = 0


        for pedido in self.pedidos:

            total += pedido.total


        return total



    def relatorio(self):

        print("\n===== RELATÓRIO =====")

        print(
            f"Produtos cadastrados: {len(self.produtos)}"
        )

        print(
            f"Clientes cadastrados: {len(self.clientes)}"
        )

        print(
            f"Pedidos realizados: {len(self.pedidos)}"
        )

        print(
            f"Faturamento: R$ {self.faturamento():.2f}"
        )

        # Menu de interação do sistema


from produto import Produto
from cliente import Cliente



class Menu:


    def _init_(self, sistema):

        self.sistema = sistema



    def executar(self):

        while True:

            print("\n==============================")
            print("   COFFEE SHOPS TIA ROSA")
            print("==============================")

            print("1 - Cadastrar produto")
            print("2 - Listar produtos")
            print("3 - Cadastrar cliente")
            print("4 - Listar clientes")
            print("5 - Fazer pedido")
            print("6 - Listar pedidos")
            print("7 - Relatório")
            print("0 - Sair")


            opcao = input("\nEscolha uma opção: ")



            if opcao == "1":

                self.cadastrar_produto()



            elif opcao == "2":

                self.sistema.listar_produtos()



            elif opcao == "3":

                self.cadastrar_cliente()



            elif opcao == "4":

                self.sistema.listar_clientes()



            elif opcao == "5":

                self.fazer_pedido()



            elif opcao == "6":

                self.sistema.listar_pedidos()



            elif opcao == "7":

                self.sistema.relatorio()



            elif opcao == "0":

                print("\nSistema encerrado.")

                break


            else:

                print("Opção inválida!")




    # -----------------------------
    # CADASTRAR PRODUTO
    # -----------------------------

    def cadastrar_produto(self):

        print("\n--- Novo Produto ---")


        codigo = int(input("Código: "))

        nome = input("Nome: ")

        categoria = input("Categoria: ")

        preco = float(input("Preço: "))

        estoque = int(input("Quantidade: "))


        produto = Produto(
            codigo,
            nome,
            categoria,
            preco,
            estoque
        )


        self.sistema.cadastrar_produto(produto)


        print("Produto cadastrado com sucesso!")




    # -----------------------------
    # CADASTRAR CLIENTE
    # -----------------------------

    def cadastrar_cliente(self):

        print("\n--- Novo Cliente ---")


        codigo = int(input("Código: "))

        nome = input("Nome: ")

        telefone = input("Telefone: ")


        cliente = Cliente(
            codigo,
            nome,
            telefone
        )


        self.sistema.cadastrar_cliente(cliente)


        print("Cliente cadastrado com sucesso!")




    # -----------------------------
    # REALIZAR PEDIDO
    # -----------------------------

    def fazer_pedido(self):


        print("\n--- Novo Pedido ---")


        codigo_cliente = int(
            input("Código do cliente: ")
        )


        cliente = self.sistema.buscar_cliente(
            codigo_cliente
        )


        if cliente is None:

            print("Cliente não encontrado!")

            return



        pedido = self.sistema.criar_pedido(cliente)



        while True:


            self.sistema.listar_produtos()


            codigo_produto = int(
                input("\nCódigo do produto (0 finalizar): ")
            )


            if codigo_produto == 0:

                break



            produto = self.sistema.buscar_produto(
                codigo_produto
            )



            if produto is None:

                print("Produto não encontrado!")

                continue



            quantidade = int(
                input("Quantidade: ")
            )



            if pedido.adicionar_produto(
                produto,
                quantidade
            ):

                print("Produto adicionado!")

            else:

                print("Estoque insuficiente!")



        print("\nPedido realizado!")

        print(pedido.exibir())
```
